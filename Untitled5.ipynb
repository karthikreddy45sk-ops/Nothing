```python
# ============================================================
# LOAN ELIGIBILITY PREDICTION USING DECISION TREE
# Dataset: dataset for loan prediction(1).csv
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files

from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report
)

# ============================================================
# STEP 1: UPLOAD DATASET
# ============================================================

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

print("\nColumns:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())


# ============================================================
# STEP 2: CHECK MISSING VALUES
# ============================================================

print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# STEP 3: DATA PREPROCESSING
# ============================================================

# Fill categorical missing values
categorical_columns = [
    "Gender",
    "Married",
    "Dependents",
    "Education",
    "Self_Employed"
]

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])


# Fill numerical missing values
numerical_columns = [
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term"
]

for column in numerical_columns:
    df[column] = df[column].fillna(df[column].median())


# ============================================================
# STEP 4: FEATURE ENGINEERING
# ============================================================

# Total income
df["TotalIncome"] = (
    df["ApplicantIncome"] +
    df["CoapplicantIncome"]
)

# Income to loan ratio
df["IncomeLoanRatio"] = (
    df["TotalIncome"] / df["LoanAmount"]
)

print("\nFeature engineering completed.")


# ============================================================
# STEP 5: CREATE PROJECT TARGET
# ============================================================

# IMPORTANT:
# The uploaded dataset does NOT contain Loan_Status.
# Therefore, this creates a project-specific eligibility label.
#
# Eligibility rule:
#   1. Total income should be >= 3000
#   2. Loan amount should be <= 40 times total income
#   3. Loan term should be <= 480 months
#
# This is NOT historical bank approval data.

df["Loan_Eligibility"] = np.where(
    (df["TotalIncome"] >= 3000) &
    (df["IncomeLoanRatio"] >= 0.025) &
    (df["Loan_Amount_Term"] <= 480),
    "Approved",
    "Rejected"
)

print("\nLoan Eligibility distribution:")
print(df["Loan_Eligibility"].value_counts())


# ============================================================
# STEP 6: SELECT INPUT FEATURES
# ============================================================

X = df[
    [
        "Gender",
        "Married",
        "Dependents",
        "Education",
        "Self_Employed",
        "ApplicantIncome",
        "CoapplicantIncome",
        "LoanAmount",
        "Loan_Amount_Term"
    ]
].copy()

y = df["Loan_Eligibility"]


# ============================================================
# STEP 7: CONVERT CATEGORICAL DATA TO NUMERICAL
# ============================================================

X = pd.get_dummies(
    X,
    columns=[
        "Gender",
        "Married",
        "Dependents",
        "Education",
        "Self_Employed"
    ],
    drop_first=True
)

print("\nFinal feature columns:")
print(X.columns.tolist())


# ============================================================
# STEP 8: TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining samples:", len(X_train))
print("Testing samples :", len(X_test))


# ============================================================
# STEP 9: CREATE DECISION TREE
# ============================================================

model = DecisionTreeClassifier(
    criterion="gini",
    max_depth=5,
    random_state=42
)


# ============================================================
# STEP 10: TRAIN MODEL
# ============================================================

model.fit(X_train, y_train)

print("\nDecision Tree training completed successfully!")


# ============================================================
# STEP 11: MAKE PREDICTIONS
# ============================================================

y_pred = model.predict(X_test)


# ============================================================
# STEP 12: MODULE 4 PERFORMANCE METRICS
# ============================================================

accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(
    y_test,
    y_pred,
    pos_label="Approved",
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    pos_label="Approved",
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_pred,
    pos_label="Approved",
    zero_division=0
)


# ============================================================
# STEP 13: DISPLAY FINAL VALUES
# ============================================================

print("\n")
print("=" * 55)
print("             MODULE 4 RESULTS")
print("=" * 55)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Accuracy  : {accuracy * 100:.2f}%")

print(f"\nPrecision : {precision:.4f}")
print(f"Precision : {precision * 100:.2f}%")

print(f"\nRecall    : {recall:.4f}")
print(f"Recall    : {recall * 100:.2f}%")

print(f"\nF1-Score  : {f1:.4f}")
print(f"F1-Score  : {f1 * 100:.2f}%")

print("=" * 55)


# ============================================================
# STEP 14: CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=["Approved", "Rejected"]
)

print("\nConfusion Matrix:")
print(cm)


# ============================================================
# STEP 15: DISPLAY CONFUSION MATRIX GRAPH
# ============================================================

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Approved", "Rejected"]
)

disp.plot()

plt.title("Loan Eligibility - Confusion Matrix")
plt.show()


# ============================================================
# STEP 16: CLASSIFICATION REPORT
# ============================================================

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        y_pred,
        labels=["Approved", "Rejected"],
        zero_division=0
    )
)


# ============================================================
# STEP 17: DECISION TREE VISUALIZATION
# ============================================================

plt.figure(figsize=(22, 12))

plot_tree(
    model,
    feature_names=X.columns,
    class_names=["Approved", "Rejected"],
    filled=True,
    rounded=True,
    fontsize=9
)

plt.title("Decision Tree - Loan Eligibility Prediction")

plt.show()


# ============================================================
# STEP 18: PERFORMANCE CHART
# ============================================================

metrics = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1-Score"
]

values = [
    accuracy,
    precision,
    recall,
    f1
]

plt.figure(figsize=(9, 6))

bars = plt.bar(metrics, values)

plt.ylim(0, 1)

plt.xlabel("Performance Metrics")
plt.ylabel("Score")

plt.title(
    "Decision Tree Model Performance"
)

for bar, value in zip(bars, values):

    plt.text(
        bar.get_x() + bar.get_width() / 2,
        value + 0.02,
        f"{value:.2f}",
        ha="center"
    )

plt.show()


# ============================================================
# STEP 19: FINAL SUMMARY FOR PPT
# ============================================================

print("\n")
print("=" * 55)
print("          VALUES TO ENTER IN MODULE 4")
print("=" * 55)

print(f"Accuracy  = {accuracy * 100:.2f}%")
print(f"Precision = {precision * 100:.2f}%")
print(f"Recall    = {recall * 100:.2f}%")
print(f"F1-Score  = {f1 * 100:.2f}%")

print("\nConfusion Matrix:")
print(cm)

print("=" * 55)


# ============================================================
# STEP 20: SAVE PREDICTIONS
# ============================================================

results = X_test.copy()

results["Actual"] = y_test.values
results["Predicted"] = y_pred

results.to_csv(
    "loan_prediction_results.csv",
    index=False
)

print("\nPrediction results saved as:")
print("loan_prediction_results.csv")
```
